## Массивы

Старый тип `array` Maple при обращении по имени может печатать только это имя. Python сразу возвращает объект массива. Для просмотра содержимого достаточно `display`; отдельного вызова `evalm` не требуется.

**Maple:** `squares:=array(1..3);`

### Трёхмерный и символьный массивы

**Maple:** `array3:=array(1..2,1..2,1..2,[[[1,2],[3,4]],[[5,6],[7,8]]]);`

```python
array3 = sp.MutableDenseNDimArray([[[1,2],[3,4]],[[5,6],[7,8]]])
prev = array3
display(prev)
```

**Maple:** `A:=array(1..2,1..2,[[a,b],[c,d]]);`

```python
A = sp.MutableDenseNDimArray([[a,b],[c,d]])
prev = A
display(prev)
```

**Maple:** `convert(A,'listlist');`

Преобразование в список списков.

```python
prev = A.tolist()
display(prev)
```

**Maple:** `convert(A,'set');`

Множество всех элементов массива.

```python
prev = {item for row in A.tolist() for item in row}
display(prev)
```

**Maple:** `convert(%,'list');`

Используется именно предыдущее множество. Сортировка нужна для устойчивого порядка печати.

```python
prev = list(sorted(prev, key=sp.default_sort_key))
display(prev)
```

### Подстановки

**Maple:** `expr:=z^2+3;`

```python
expr = z**2+3
prev = expr
display(prev)
```

**Maple:** `subs({z=x+y},expr);`

Функция замены (подстановки).

```python
prev = expr.subs({z:x+y})
display(prev)
```

**Maple:** `subs({2=9},pwrs);`

В исходном Maple эта команда действует на имя старого массива, а не его элементы: результат — `pwrs`. В Python имя автоматически раскрывается в объект; чтобы показать именно смысл этого примера, подстановка здесь применяется к символу имени.

```python
prev = sp.Symbol('pwrs').subs({sp.Integer(2):sp.Integer(9)})
display(prev)
```

**Функция замены — другой комментарий** (текст из исходной лекции).

**Maple:** `subs({2=9},evalm(pwrs));`

Теперь подстановка применяется к элементам вычисленного массива. Возвращается новый массив, исходный `pwrs` не изменяется.

```python
prev = pwrs.applyfunc(lambda item: sp.sympify(item).subs({2:9}))
display(prev)
```

**Maple:** `evalm(pwrs);`

Проверка: в исходном массиве по-прежнему находится 2.

```python
prev = pwrs
display(prev)
```

## Таблицы

Аналог Maple `table` — словарь `dict`. В исходнике ключи и переводы являются именами, поэтому они представлены символами SymPy, а не строками. Порядок отображения ключей не влияет на обращение к таблице.

**Maple:** `dictionary:=table([one=один,two=два,three=три]);`

```python
dictionary = {one:один, two:два, three:три}
prev = dictionary
display(prev)
```

**Maple:** `dictionary[two];`

```python
prev = dictionary[two]
display(prev)
```

**Maple:** `earth_data:=table([mass=[5.976*10^24,kg],radius=[6.378164*10^6,m],circumference=[4.00752*10^7,m]]);`

Десятичные значения сохранены с точностью Maple по умолчанию — 10 значащих цифр. `kg` и `m` — символьные обозначения единиц из лекции, без отдельной системы перевода единиц.

```python
earth_data = {
    mass: [sp.Float('5.976', Digits)*10**24, kg],
    radius: [sp.Float('6.378164', Digits)*10**6, m],
    circumference: [sp.Float('4.00752', Digits)*10**7, m],
}
prev = earth_data
display(prev)
```

**Maple:** `earth_data[mass];`

```python
prev = earth_data[mass]
display(prev)
```
